## 5.1 Functions: Concepts, Pseudocode, and Flowchart

A **function** is a named, reusable block of code that performs one specific task. Instead of writing the same calculation many times, you define it once and **call** it whenever needed. In engineering, functions mirror physical subsystems: a stress calculator, a unit converter, or a safety-factor checker each does one well-defined job.

Before writing code, use **pseudocode** to define the function's interface and algorithm:
```
FUNCTION function_name(parameter1, parameter2)
    algorithm steps
    RETURN result
END FUNCTION
```

Python's `def` keyword translates this directly.


````python
# Worked example — pseudocode → Python: Celsius to Fahrenheit
# Pseudocode:
#   FUNCTION celsius_to_fahrenheit(celsius)
#       SET fahrenheit = (celsius * 9/5) + 32
#       RETURN fahrenheit
#   END FUNCTION

def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

# Call the function for several temperatures
test_temps_C = [0, 100, -40, 37]
for temp_C in test_temps_C:
    temp_F = celsius_to_fahrenheit(temp_C)
    print(f"{temp_C:>5} °C  →  {temp_F:>7.2f} °F")
````

## 5.2 Defining and Calling Programmer-Defined Functions

The `def` keyword creates a new function. The function must be **defined before it is called**. The syntax is:

```python
def function_name(parameter1, parameter2):
    statement(s)          # body — indented 4 spaces
    return value          # optional
```

- **Parameters** are the variable names listed inside the parentheses in the definition.
- **Arguments** are the actual values passed when you call the function.
- A **void function** performs an action (e.g., printing) without returning a value — omit the `return` statement.


````python
# Worked example — calculate axial strain from stress and modulus
# strain = stress_MPa / (E_GPa * 1000)   (unit conversion: GPa → MPa)

def calculate_strain(stress_MPa, E_GPa):
    strain = stress_MPa / (E_GPa * 1000)   # dimensionless
    return strain

# Call the function for three materials
print(f"{'Material':<15} {'Stress (MPa)':<15} {'E (GPa)':<10} {'Strain':<12}")
print("-" * 52)

data = [("Steel",    200, 200),
        ("Aluminum", 120,  70),
        ("Titanium", 600, 110)]

for material, stress, E in data:
    strain = calculate_strain(stress, E)
    print(f"{material:<15} {stress:<15} {E:<10} {strain:<12.6f}")
````

# 1.3 - Handling Error

Error message is important information to identify mistakes in our program. Three basic error;

1. Runtime errors: Errors where syntactically valid code fails to execute, perhaps due to invalid user input.
2. Syntax errors: Errors where the code is not valid Python Syntax.
3. Semantic errors: Errors in logic: code executes without a problem, but the result is not what you expect. (Difficult to detect)
